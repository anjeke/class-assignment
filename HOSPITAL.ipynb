```python
class Patient:
    def __init__(self,patients_id,name,age,gender,disease):
        self.patients_id = patients_id
        self.name = name
        self.age = age
        self.gender = gender
        self.disease = disease
    def display(self):
        print("---Patients details----")
        print(f"Patients_id:{self.patients_id}")
        print(f"Name:{self.name}")
        print(f"Age:{self.age}")
        print(f"Gender:{self.gender}")
        print(f"Disease:{self.disease}")
patients=[]
class doctor:
    def __init__(self,doctors_id,name,age,speciality):
        self.doctors_id = doctors_id
        self.name = name
        self.age= age
        self.speciality = speciality
    def display(self):
        print("---Doctors details----")
        print(f"doctors_id:{self.doctors_id}")
        print(f"Name:{self.name}")
        print(f"Age:{self.age}")
        print(f"speciality:{self.speciality}")

doctors=[]
# Patients_details 

Patients_id=input("Enter the patients ID:")
Patient_Name=input("Enter the name: ")
Patient_Age=int(input("Enter the age:"))
Gender=input("Enter the patients gender")
Disease=input("Enter the patients disease:")

X1=Patient(Patients_id,Patient_Name,Patient_Age,Gender,Disease)
patients.append(X1)

print("\n")
# Doctors_details
doctors_id=input("Enter the doctorsID:")
Doctor_Name=input("Enter the name: ")
Doctor_Age=int(input("Enter the age:"))
Speciality=input("Enter the doctors speciality")

Doctor1=doctor(doctors_id,Doctor_Name,Doctor_Age,Speciality)
doctors.append(Doctor1)


while True:
    print("\n===== Hospital Management System =====")
    print("1. Add patient")
    print("2. display patient")
    print("3. Add doctor")
    print("4.Display doctor")
    print("5. Exit")

    choice=input("Enter your choice: ")
    if choice =="1":
       Patients_id=input("Enter the patients ID:")
       Patient_Name=input("Enter the name: ")
       Patient_Age=int(input("Enter the age:"))
       Gender=input("Enter the patients gender")
       Disease=input("Enter the patients disease:")

       X1=Patient(Patients_id,Patient_Name,Patient_Age,Gender,Disease)
       patients.append(X1)
       

       print("Patient Added succesfully!")

    elif choice =="2":
         if len(patients)==0:
            print("No patients found")
         else:
             print("---Patients details----")
             for patient in patients:
                patient.display()


    elif choice == "3":
       doctors_id=input("Enter the doctorsID:")
       Doctor_Name=input("Enter the name: ")
       Doctor_Age=int(input("Enter the age:"))
       Speciality=input("Enter the doctors speciality")

       Doctor1=doctor(doctors_id,Doctor_Name,Doctor_Age,Speciality)
       doctors.append(Doctor1)

       print("Doctor added succesfully!")
    elif choice == "4":
        if len(doctors)==0:
            print("No doctor found")
        else:
             print("---doctors  details----")
             for doctor in doctors:
                doctor.display()
    elif choice == "5":
       print("Thank you.Welcome again")
       break
    else:
       print("invalid choice entered.Please try again")
```
